```python


import csv
import os




ORDER_FILE = "orders.csv"




products = [
    {
        "id": 101,
        "name": "Classic T-Shirt",
        "category": "T-Shirt",
        "sizes": ("S", "M", "L", "XL"),
        "price": 499.00,
        "stock": 10
    },
    {
        "id": 102,
        "name": "Denim Jacket",
        "category": "Jacket",
        "sizes": ("M", "L", "XL"),
        "price": 1499.00,
        "stock": 5
    },
    {
        "id": 103,
        "name": "Floral Dress",
        "category": "Dress",
        "sizes": ("S", "M", "L"),
        "price": 1299.00,
        "stock": 7
    },
    {
        "id": 104,
        "name": "Casual Hoodie",
        "category": "Hoodie",
        "sizes": ("M", "L", "XL"),
        "price": 999.00,
        "stock": 8
    },
    {
        "id": 105,
        "name": "Formal Shirt",
        "category": "Shirt",
        "sizes": ("S", "M", "L", "XL"),
        "price": 899.00,
        "stock": 12
    },
    {
        "id": 106,
        "name": "Wide Leg Jeans",
        "category": "Jeans",
        "sizes": ("S", "M", "L"),
        "price": 1199.00,
        "stock": 6
    }
]




def create_order_file():

    if not os.path.exists(ORDER_FILE):

        try:

            with open(
                ORDER_FILE,
                "w",
                newline=""
            ) as file:

                writer = csv.writer(file)

                writer.writerow([
                    "Customer",
                    "Product",
                    "Size",
                    "Quantity",
                    "Amount"
                ])

        except Exception:

            print("Error creating order file.")




class ShoppingCart:



    def __init__(self):

        self.cart = []

        self.wishlist = set()

        self.discount = 0




    def view_products(self):

        print("\n========== FASHION PRODUCTS ==========")

        for product in products:

            print(
                f"ID: {product['id']} | "
                f"{product['name']} | "
                f"{product['category']} | "
                f"₹{product['price']:.2f} | "
                f"Stock: {product['stock']}"
            )

            print(
                "Available Sizes:",
                ", ".join(product["sizes"])
            )

            print("-" * 50)



    def search_product(self):

        keyword = input(
            "\nEnter product name to search: "
        ).strip().lower()

        if keyword == "":

            print("Search term cannot be empty.")

            return

        found = False

        for product in products:

            if keyword in product["name"].lower():

                print("\nProduct Found!")

                print(
                    f"ID       : {product['id']}"
                )

                print(
                    f"Name     : {product['name']}"
                )

                print(
                    f"Category : {product['category']}"
                )

                print(
                    f"Price    : ₹{product['price']:.2f}"
                )

                print(
                    f"Stock    : {product['stock']}"
                )

                print(
                    "Sizes    :",
                    ", ".join(product["sizes"])
                )

                found = True

        if not found:

            print("No matching product found.")



    def add_to_cart(self):

        self.view_products()

        try:

            product_id = int(
                input("\nEnter Product ID: ")
            )

            product = None

            for item in products:

                if item["id"] == product_id:

                    product = item

                    break

            if product is None:

                print("Product not found.")

                return

            if product["stock"] <= 0:

                print("This product is out of stock.")

                return

            print(
                "\nAvailable Sizes:",
                ", ".join(product["sizes"])
            )

            size = input(
                "Enter Size: "
            ).strip().upper()

            if size not in product["sizes"]:

                print(
                    "Invalid size. "
                    "Please choose an available size."
                )

                return

            quantity = int(
                input("Enter Quantity: ")
            )

            if quantity <= 0:

                print(
                    "Quantity must be greater than zero."
                )

                return

            if quantity > product["stock"]:

                print(
                    f"Only {product['stock']} "
                    "items are available."
                )

                return



            for item in self.cart:

                if (
                    item["id"] == product["id"]
                    and item["size"] == size
                ):

                    new_quantity = (
                        item["quantity"] + quantity
                    )

                    if new_quantity > product["stock"]:

                        print(
                            "Requested quantity exceeds stock."
                        )

                        return

                    item["quantity"] = new_quantity

                    print(
                        "Cart quantity updated successfully!"
                    )

                    return



            self.cart.append({
                "id": product["id"],
                "name": product["name"],
                "size": size,
                "price": product["price"],
                "quantity": quantity
            })

            print(
                product["name"],
                "added to cart successfully!"
            )

        except ValueError:

            print(
                "Invalid input! "
                "Please enter a valid number."
            )



    def calculate_subtotal(self):

        subtotal = 0

        for item in self.cart:

            subtotal += (
                item["price"]
                * item["quantity"]
            )

        return subtotal




    def calculate_total(self):

        subtotal = self.calculate_subtotal()

        discount_amount = (
            subtotal * self.discount / 100
        )

        total = (
            subtotal - discount_amount
        )

        return total




    def view_cart(self):

        print("\n========== YOUR SHOPPING CART ==========")

        if len(self.cart) == 0:

            print("Your cart is empty.")

            return

        for number, item in enumerate(
            self.cart,
            start=1
        ):

            amount = (
                item["price"]
                * item["quantity"]
            )

            print(
                f"{number}. "
                f"{item['name']} | "
                f"Size: {item['size']} | "
                f"Quantity: {item['quantity']} | "
                f"₹{amount:.2f}"
            )

        print("-" * 50)

        subtotal = self.calculate_subtotal()

        discount_amount = (
            subtotal * self.discount / 100
        )

        total = self.calculate_total()

        print(
            f"Subtotal        : ₹{subtotal:.2f}"
        )

        print(
            f"Discount        : {self.discount}%"
        )

        print(
            f"Discount Amount : ₹{discount_amount:.2f}"
        )

        print(
            f"Final Total     : ₹{total:.2f}"
        )



    def remove_from_cart(self):

        if len(self.cart) == 0:

            print("Your cart is empty.")

            return

        self.view_cart()

        try:

            number = int(
                input(
                    "\nEnter item number to remove: "
                )
            )

            if 1 <= number <= len(self.cart):

                removed = self.cart.pop(
                    number - 1
                )

                print(
                    removed["name"],
                    "removed from cart."
                )

            else:

                print("Invalid item number.")

        except ValueError:

            print(
                "Invalid input! "
                "Please enter a valid number."
            )



    def update_quantity(self):

        if len(self.cart) == 0:

            print("Your cart is empty.")

            return

        self.view_cart()

        try:

            number = int(
                input(
                    "\nEnter item number to update: "
                )
            )

            if not (
                1 <= number <= len(self.cart)
            ):

                print("Invalid item number.")

                return

            quantity = int(
                input(
                    "Enter new quantity: "
                )
            )

            if quantity <= 0:

                print(
                    "Quantity must be greater than zero."
                )

                return

            item = self.cart[
                number - 1
            ]

            product = None

            for product_item in products:

                if (
                    product_item["id"]
                    == item["id"]
                ):

                    product = product_item

                    break

            if product is None:

                print("Product not found.")

                return

            if quantity > product["stock"]:

                print(
                    f"Only {product['stock']} "
                    "items are available."
                )

                return

            item["quantity"] = quantity

            print(
                "Quantity updated successfully!"
            )

        except ValueError:

            print(
                "Invalid input! "
                "Please enter a valid number."
            )



    def apply_discount(self):

        if len(self.cart) == 0:

            print(
                "Add products to your cart first."
            )

            return

        print("\n========== DISCOUNT OPTIONS ==========")

        print("1. SAVE10 - 10% discount")

        print("2. SAVE20 - 20% discount")

        print("3. Remove Discount")

        choice = input(
            "Enter your choice: "
        ).strip()

        if choice == "1":

            self.discount = 10

            print(
                "10% discount applied successfully!"
            )

        elif choice == "2":

            self.discount = 20

            print(
                "20% discount applied successfully!"
            )

        elif choice == "3":

            self.discount = 0

            print(
                "Discount removed."
            )

        else:

            print("Invalid discount choice.")



    def add_to_wishlist(self):

        self.view_products()

        try:

            product_id = int(
                input(
                    "\nEnter Product ID: "
                )
            )

            found = False

            for product in products:

                if product["id"] == product_id:

                    self.wishlist.add(
                        product_id
                    )

                    print(
                        product["name"],
                        "added to wishlist!"
                    )

                    found = True

                    break

            if not found:

                print("Product not found.")

        except ValueError:

            print(
                "Invalid Product ID."
            )




    def view_wishlist(self):

        print("\n========== YOUR WISHLIST ==========")

        if len(self.wishlist) == 0:

            print("Your wishlist is empty.")

            return

        for product in products:

            if product["id"] in self.wishlist:

                print(
                    f"ID: {product['id']} | "
                    f"{product['name']} | "
                    f"₹{product['price']:.2f}"
                )



    def checkout(self):

        if len(self.cart) == 0:

            print(
                "Your cart is empty."
            )

            return

        customer = input(
            "\nEnter Customer Name: "
        ).strip()

        if customer == "":

            customer = "Customer"

        self.view_cart()

        confirmation = input(
            "\nConfirm order? (yes/no): "
        ).strip().lower()

        if confirmation != "yes":

            print(
                "Order cancelled."
            )

            return

        try:

            subtotal = self.calculate_subtotal()

            discount_amount = (
                subtotal
                * self.discount
                / 100
            )

            total = (
                subtotal
                - discount_amount
            )

            with open(
                ORDER_FILE,
                "a",
                newline=""
            ) as file:

                writer = csv.writer(file)

                for item in self.cart:

                    amount = (
                        item["price"]
                        * item["quantity"]
                    )

                    writer.writerow([
                        customer,
                        item["name"],
                        item["size"],
                        item["quantity"],
                        amount
                    ])


            for item in self.cart:

                for product in products:

                    if (
                        product["id"]
                        == item["id"]
                    ):

                        product["stock"] -= (
                            item["quantity"]
                        )

            print("\n========== ORDER CONFIRMED ==========")

            print(
                f"Customer : {customer}"
            )

            print(
                f"Subtotal : ₹{subtotal:.2f}"
            )

            print(
                f"Discount : ₹{discount_amount:.2f}"
            )

            print(
                f"Total Paid : ₹{total:.2f}"
            )

            print(
                "Order saved successfully!"
            )

            print(
                "Thank you for shopping with us!"
            )



            self.cart.clear()

            self.discount = 0

        except Exception:

            print(
                "Error while saving the order."
            )



def view_orders():

    print("\n========== PREVIOUS ORDERS ==========")

    try:

        with open(
            ORDER_FILE,
            "r"
        ) as file:

            reader = csv.DictReader(file)

            orders = list(reader)

            if len(orders) == 0:

                print(
                    "No previous orders found."
                )

                return

            for number, order in enumerate(
                orders,
                start=1
            ):

                print(
                    f"{number}. "
                    f"Customer: {order['Customer']} | "
                    f"Product: {order['Product']} | "
                    f"Size: {order['Size']} | "
                    f"Quantity: {order['Quantity']} | "
                    f"Amount: ₹{float(order['Amount']):.2f}"
                )

    except FileNotFoundError:

        print(
            "Order file not found."
        )

    except Exception:

        print(
            "Error reading order history."
        )



def display_menu():

    print("\n" + "=" * 60)

    print(
        "           🛍️ FASHION SHOPPING SYSTEM"
    )

    print("=" * 60)

    print("1. View Products")

    print("2. Search Product")

    print("3. Add Product to Cart")

    print("4. View Cart")

    print("5. Update Cart Quantity")

    print("6. Remove Product from Cart")

    print("7. Apply Discount")

    print("8. Add Product to Wishlist")

    print("9. View Wishlist")

    print("10. Checkout")

    print("11. View Previous Orders")

    print("12. Exit")

    print("=" * 60)




create_order_file()

cart = ShoppingCart()


while True:

    display_menu()

    choice = input(
        "Enter your choice (1-12): "
    ).strip()


    if choice == "1":

        cart.view_products()


    elif choice == "2":

        cart.search_product()


    elif choice == "3":

        cart.add_to_cart()


    elif choice == "4":

        cart.view_cart()


    elif choice == "5":

        cart.update_quantity()


    elif choice == "6":

        cart.remove_from_cart()


    elif choice == "7":

        cart.apply_discount()


    elif choice == "8":

        cart.add_to_wishlist()


    elif choice == "9":

        cart.view_wishlist()


    elif choice == "10":

        cart.checkout()


    elif choice == "11":

        view_orders()


    elif choice == "12":

        print(
            "\nThank you for using "
            "Fashion Shopping System!"
        )

        print(
            "Program ended successfully."
        )

        break


    else:

        print(
            "\nInvalid choice!"
        )

        print(
            "Please enter a number from 1 to 12."
        )
```
